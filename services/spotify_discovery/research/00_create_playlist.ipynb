```python
import spotipy
from spotipy.oauth2 import SpotifyOAuth

scope = "user-library-read"

sp = spotipy.Spotify(auth_manager=SpotifyOAuth(scope=scope))

results = sp.current_user_saved_tracks()
for idx, item in enumerate(results['items']):
    track = item['track']
    print(idx, track['artists'][0]['name'], " – ", track['name'])
```

```python
# i want to create a new playlist with the songs in song_urls
scope = "playlist-modify-private"

sp = spotipy.Spotify(auth_manager=SpotifyOAuth(scope=scope))

song_urls = [
    "https://open.spotify.com/intl-de/track/2CoWJhDYVENsByS7rPInSr?si=3aa61d2d054148b8",
    "https://open.spotify.com/intl-de/track/1kQhi22kyluRCnIPAXtxEZ?si=1a8f0c1361a1419e",
]

playlist = sp.user_playlist_create(user=sp.current_user()['id'], name="My New Playlist", public=False)
sp.playlist_add_items(playlist_id=playlist['id'], items=song_urls)
```
