# Rekordbox readable export

this script will create an excel file based on the latest rekordbox export (this helps to check if a track is already in the library)

this function assumes, that the rekordbox database has all the xml files in the following format: `rekordbox7_YYYY-MM-DD.xml`

```python
import os
os.chdir('..')
print(os.getcwd())
```

```python
from dotenv import load_dotenv
environment = "local"
print("successfully loaded env file: ", load_dotenv(f"./env/.env.{environment}", override=True))
```

```python
import rekordbox_client.dataloading as dataloading
```

```python
database_folder = os.environ["DATABASE_FOLDER"]
```

```python
# find the most recent rekordbox7_YYYY-MM-DD.xml file
latest = sorted(f for f in os.listdir(database_folder) if f.startswith("rekordbox7_") and f.endswith(".xml"))[-1]

df = dataloading.load_dataframe_from_rekordbox_xml(os.path.join(database_folder, latest))
df.head()
```

```python
latest_filenamesplit = latest.split(".")[0].split("_")
outfilename = latest_filenamesplit[0] + '_simple_' + latest_filenamesplit[-1] + '.xlsx'
print(outfilename)
```

```python
output_folder = os.environ["OUTPUT_FOLDER"]

df[["@Name", "@Artist"]].sort_values(
    by=[
        "@Name",
        "@Artist",
    ]
).to_excel(os.path.join(output_folder, outfilename), index=False)
```
